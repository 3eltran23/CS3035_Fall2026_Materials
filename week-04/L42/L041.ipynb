```python
import sys


def get_int(prompt):
    while True:
        try:
            x = int(input(prompt))
        except ValueError:
            print("x is not an integer")
        else:
            return x


def show_menu():
    print("\n" + "=" * 20)
    print("      MAIN MENU")
    print("=" * 20)
    print("1. Greet User")
    print("2. Display System Info")
    print("3. Exit")
    print("=" * 20)


def input_choice():
    while True:
        show_menu()
        choice = get_int("Enter your choice (1-3): ")

        match choice:
            case 1:
                print("\n--> Hello! Welcome to the Python menu program.")
            case 2:
                print(f"\n--> Python Version: {sys.version}")
            case 3:
                print("\nExiting program. Goodbye!")
                break
            case _:
                print("\n[!] Invalid choice. Please enter 1, 2, or 3.")


if __name__ == "__main__":
    input_choice()
```
