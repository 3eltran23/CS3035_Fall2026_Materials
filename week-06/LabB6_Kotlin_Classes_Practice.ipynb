## 2. Properties and the primary constructor

Properties describe an object's state. Put constructor properties in the class header:

- `val` creates a read-only property;
- `var` creates a property that can be reassigned.

Prefer `val` unless the value genuinely needs to change. A constructor parameter may also have a default value. Properties can be initialized in the class body as well.

```kotlin
class Contact(
    val id: Int,
    var email: String = "unknown@example.com"
) {
    val category: String = "school"

    fun printSummary() {
        println("Contact #$id: $email [$category]")
    }
}
```

## 3. Creating instances and using properties

Calling the class name like a function invokes its constructor. Use dot notation to read a property, update a `var`, or call a member function.

```kotlin
val courseContact = Contact(101, "instructor@example.edu")

println(courseContact.email)
courseContact.email = "cs3035@example.edu"
println("Updated email: ${courseContact.email}")
courseContact.printSummary()
```

## 4. Member functions

A function declared inside a class describes behavior shared by its instances. It can refer directly to the instance's properties.

```kotlin
class Counter(val label: String, var value: Int = 0) {
    fun increment() {
        value += 1
    }

    fun display() {
        println("$label: $value")
    }
}

val attendance = Counter("Attendance")
attendance.increment()
attendance.increment()
attendance.display()
```

## 5. Data classes

Use `data class` for objects whose main purpose is to store values. Kotlin generates useful behavior from the properties in the primary constructor, including:

- a readable `toString()` representation;
- value-based equality with `==`; and
- `copy()` for creating a modified copy without changing the original.

Properties declared only in the class body are not included in those generated operations.

```kotlin
data class Student(val name: String, val id: Int)

val studentA = Student("Alex", 1)
val studentB = Student("Alex", 1)
val studentC = Student("Mina", 2)

println(studentA)
println("studentA == studentB: ${studentA == studentB}")
println("studentA == studentC: ${studentA == studentC}")
```

### Copying a data-class instance

Named arguments make it clear which property changes. The original object is left unchanged.

```kotlin
val renumberedStudent = studentA.copy(id = 99)

println("Original: $studentA")
println("Copy:     $renumberedStudent")
```
