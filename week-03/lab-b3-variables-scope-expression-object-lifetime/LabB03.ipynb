## Before you begin: Equality and object identity

| Question | Python | Kotlin |
|---|---|---|
| Equal values or contents? | `a == b` | `a == b` |
| Same object or reference? | `a is b` | `a === b` |
| Different values or contents? | `a != b` | `a != b` |
| Different objects or references? | `a is not b` | `a !== b` |

Equal contents do not necessarily mean that two variables refer to the same object. Two variables that refer to the same object are aliases. In Python, use `is` for identity checks such as `value is None`, not for comparing string or numeric values.

````python
first = [10, 20]
second = [10, 20]
alias = first

print(f"Equal contents: {first == second}")
print(f"Same object, first and second: {first is second}")
print(f"Same object, first and alias: {first is alias}")
````
