```python
import random


def main():
    first_result = random.randint(1, 6)
    second_result = random.randint(1, 6)

    if first_result == second_result:
        print("You win :)")
    else:
        print("You lose :(")


if __name__ == "__main__":
    main()
```

## Exercise A2.4 - Control Flow: Traffic Light


```python
def main():
    traffic_light_state = "Red"

    match traffic_light_state:
        case "Green":
            traffic_action = "Go"
        case "Yellow":
            traffic_action = "Slow down"
        case "Red":
            traffic_action = "Stop"
        case _:
            traffic_action = "Malfunction"

    print(traffic_action)


if __name__ == "__main__":
    main()
```

## Exercise A2.5 - Loops: Pizza Slices


```python
def main():
    pizza_slices = 0

    while pizza_slices < 8:
        pizza_slices += 1

        if pizza_slices < 8:
            print(f"There's only {pizza_slices} slice/s of pizza :(")

    print(
        f"There are {pizza_slices} slices of pizza. "
        "Hooray! We have a whole pizza! :D"
    )


if __name__ == "__main__":
    main()
```

## Exercise A2.6 - Loops: FizzBuzz
Write a program that simulates the [Fizz buzz game](https://en.wikipedia.org/wiki/Fizz_buzz). Your task is to print numbers from 1 to 100 incrementally, replacing any number divisible by three with the word "fizz", and any number divisible by five with the word "buzz". Any number divisible by both 3 and 5 must be replaced with the word "fizzbuzz".



```python
def main():
    for number in range(1, 101):
        if number % 15 == 0:
            print("fizzbuzz")
        elif number % 3 == 0:
            print("fizz")
        elif number % 5 == 0:
            print("buzz")
        else:
            print(number)


if __name__ == "__main__":
    main()
```

## Exercise A2.7 - Loops: Words Beginning with “l”

You have a list of words. Use  `for` and `if` to print only the words that start with the letter l.


```python
def main():
    words = ["dinosaur", "limousine", "magazine", "language"]

    for word in words:
        if word.startswith("l"):
            print(f"{word}")


if __name__ == "__main__":
    main()
```

## Exercise A2.12 - Functions: Default and Named Arguments


```python
def interval_in_seconds(
    hours: int = 0,
    minutes: int = 0,
    seconds: int = 0,
) -> int:
    return ((hours * 60) + minutes) * 60 + seconds


def main():
    print(interval_in_seconds(1, 20, 15))
    print(interval_in_seconds(minutes=1, seconds=25))
    print(interval_in_seconds(hours=2))
    print(interval_in_seconds(minutes=10))
    print(interval_in_seconds(hours=1, seconds=1))


if __name__ == "__main__":
    main()
```
