## What is a program?
A program is a sequence of instructions that specifies how to perform a computation. The computation might be something mathematical, such as solving a system of equations or finding the roots of a polynomial, but it can also be a symbolic computation, such as searching and replacing text in a document.

The details look different in different languages, but a few basic instructions appear in just about every language:

* **input:** Get data from the keyboard, a file, or some other device.
* **output:** Display data on the screen or send data to a file or other device.
* **math:** Perform basic mathematical operations like addition and multiplication.
* **conditional execution:** Check for certain conditions and execute the appropriate code.
* **repetition:** Perform some action repeatedly, usually with some variation.

* The general syntax for list slicing is `list[start:stop:step]`, where:

* `start`:

  * The index of the first element to include in the slice `(inclusive)`. If omitted, it defaults to the beginning of the list `(index 0)`.

* `stop`:

  * The index of the element before which the slice should end `(exclusive)`. If omitted, it defaults to the end of the list.

* `step`:

  * The number of elements to skip between each included element. If omitted, it `defaults` to `1`. A `negative step` value reverses the order of the slice.

#### Conditional Statements

Structure

```
if <expression>:
    <suite>
elif <expression>:
    <suite>**bold text**
else:
    <suite>
```
* if → first condition.

* elif → check more conditions (optional, can be multiple).

* else → run if none of the above are true (optional).

* Rules:

  * Each header ends with a colon (:).

  * Suites are indented consistently.

  * No parentheses needed around conditions.




````python
if 2**3 > 5:
    print("greater than")
elif 2**3 == 5:
    print("equal")
else:
    print("less than")
````

````python
d = 0
check = True

if check:
    d = 1
else:
    d = 2


print(d)
# 1
````

### Match
A match statement takes an expression and compares its value to successive patterns given as one or more case blocks. This is superficially similar to a switch statement in C, Java or JavaScript (and many other languages), but it’s more similar to pattern matching in languages like Rust or Haskell. Only the first pattern that matches gets executed and it can also extract components (sequence elements or object attributes) from the value into variables. If no case matches, none of the branches is executed.

````python
def http_error(status):
    match status:
        case 400:
            return "Bad request"
        case 404:
            return "Not found"
        case 418:
            return "I'm a teapot"
        case _:
            return "Something's wrong with the internet"


http_error(404)
````

An expression returns a value that can be used later in your code.

````python
obj = "Hell"

match obj:
    case "1":
        print("One")
    case "Hello":
        print("Greetings")
    case _:
        print("Unknown")
````

## Functions
You can declare your own functions in python using the `def` keyword.



````python
def hello():
    print("Hello, world!")


if __name__ == "__main__":
    hello()
````

````python
def square(x):
    return x * x


def print_twice(s):
    print(s)
    print(s)


x = print_twice("Hello")
print(x)
````

In Python:
- Functions are declared using the def keyword.
- Function parameters are written within parentheses ().
- Parameter type annotations are optional. Multiple parameters are separated by commas ,.
- An optional return-type annotation is written after the parentheses using ->.
- A colon : marks the beginning of the function body.
- The function body is defined using indentation instead of curly braces.
- The return keyword exits the function and can return a value.
- A function without an explicit return statement returns None.

````python
def sum(x, y):
    return x + y


print(sum(3, 4))  # 7
````

````python
def sum(x: int, y: int) -> int:
    return x + y


print(sum(3, 4))  # 7
````

### Default Parameter Values
You can define default values for Python function parameters. A parameter with a default value can be omitted when the function is called.
Use the assignment operator = after the parameter name or its optional type annotation:

````python
def print_message_with_prefix(
    message: str,
    prefix: str = "Info",
):
    print(f"[{prefix}] {message}")
````

````python
 print_message_with_prefix("Hello", "Log")
````

````python
print_message_with_prefix("Hello")
````

````python
 print_message_with_prefix(prefix="Log", message="Hello")
````
