# Functions as Values

- Kotlin allows **functions to be passed as values** to other functions.
- Functions can be passed:
  - By **reference** (named functions)
  - **Anonymously** (lambdas / function literals)
- This style is familiar to programmers using **Java 8+ functional features**.

- **Function references**:
  - Use `::functionName` to pass an existing function
  - Examples:
    - `::abs` for functions in the same scope
    - `Example::abs` for functions in a companion object
  - Imported functions can be referenced directly:
    - `formatResult("factorial", 7, ::factorial)`

## Lambda Expressions

- Kotlin allows you write even more concise code for functions by using lambda expressions

- **Lambda expressions (lambdas)** are small blocks of code that can be passed as values to other functions
- **Lambdas** are commonly used with higher-order functions

````kotlin
fun uppercaseString(text: String): String {
    return text.uppercase()
}

print(uppercaseString("hello")) //HELLO
````

can also be written as a **lambda** expression

````kotlin
val upperCaseString = { text: String -> text.uppercase() }
println(upperCaseString("hello"))
````

Passing and sotring pieces of behavior in your code is a frequent task. You ofter need to express idea like, “Apply this operation to all elements in a data structure.”

> Lambdas allows to pass behavior into functions.

Instead of writing loops, you describe an operation once and apply it everywhere. This idea is a common across programming languages and is heavily used in Kotlin's library

### Understanding Lambda Expressions (Kotlin)

Similar to other concepts in programming languages, such as regular expressions, lambda expressions can be hard to understand at frist glance. This is mostly because they introduce a compact syntax that looks unfamiliar at frist.

So, let's break it down.

In Kotlin, a **lambda expression** is an anonymous function written inside curly braces `{}`. It represents a block of code that can be passed as a value, stored in a variable, or provided as an argument to another function.


Basic Syntax

```kotlin
{ parameters -> body }
```

- **Parameters** appear on the left side of `->`

- The **function body** appears on the right after the `->`

- The value of the last expression is automatically returned

> If you declare a lambda without parameters, then there is no need to use `->`. For example

````kotlin
val helloString = { println("Hello world!")}
helloString()
````

### Pass to another function (Kotlin)

A great example of when it is useful to pass a lambda expression to a function is when using the `filter()` function on collections.

The `filter()` function is a higher-order function, meaning it takes another function (a lambda) as an argument. This lambda defines the condition that each element in the collection must satisfy in order to be included in the result.

#### Example

````kotlin
val numbers = listOf(1, -2, 3, -4, 5, -6)

val positives = numbers.filter ({ x -> x > 0 })

val isNegative = { x: Int -> x < 0 }
val negatives = numbers.filter(isNegative)

println("positives: ${positives}")
// [1, 3, 5]
println("negatives: ${negatives}")
// [-2, -4, -6]
````

The function keeps an element only if the predicate returns **true**:

- `{ x -> x > 0 }` returns `true` if the element is **positive**
- `{ x -> x < 0 }` returns `true` if the element is **negative**


> If a lambda expression is the only function parameter, you can drop the function parentheses `()`:

````kotlin
val numbers = listOf(1, -2, 3, -4, 5, -6)

val positives = numbers.filter { x -> x > 0 }

val isNegative = { x: Int -> x < 0 }
val negatives = numbers.filter(isNegative)

println("positives: ${positives}")
// [1, 3, 5]
````

### Return from a function

Lambda expressions can also be **returned from a function**.

In order for hte compiler to understand the type of the returned lambda, the function must declare a **function type** as its return


In the following example, the `toSeconds()` function has the function type **`(Int) -> Int`** because it always returns a lambda expression that:

- takes one parameter of type `Int`
- returns a value of type `Int`

This example uses a `when` expression to determine which lambda expression is returned when `toSeconds()` is called.


````kotlin
fun toSeconds(time: String): (Int) -> Int = when (time) {
    "hour" -> { value -> value * 60 * 60 }
    "minute" -> { value -> value * 60 }
    "second" -> { value -> value }
    else -> { value -> value }
}



fun main() {
    val timesInMinutes = listOf(2, 10, 15, 1)
    val min2sec = toSeconds("minute")
    val totalTimeInSeconds = timesInMinutes.map(min2sec).sum()
    println("Total time is $totalTimeInSeconds secs")
    // Total time is 1680 secs
}

main()
````

> **Reminder:** `toSeconds` is a *single-expression function*, meaning its entire body consists of a single expression whose value is automatically returned.

### Invoke separately

Lambda expressions can be invoked on their own by adding parentheses `()` after the curly braces `{}` and including any required parameters inside the parentheses.

#### Example
```kotlin
{ println("Hello world!") }()
```

In this example:

- `{ println("Hello world!") }` defines a lambda expression

- `()` immediately invokes the lambda

- The code inside the lambda is executed right away


````kotlin
println({ text: String -> text.uppercase() }("hello"))
// HELLO
````

## Trailing Lambdas

As you have already seen, if a lambda expression is the **only parameter** of a function, you can drop the function parentheses `()`.
If a lambda expression is passed as the **last parameter** of a function, the lambda can be written **outside** the parentheses.

In both cases, this syntax is called a **trailing lambda**.

---

### Example: `fold()`

The `fold()` function accepts:
- an **initial value**
- an **operation** (a lambda)


In Kotlin, **`fold`** is conceptually similar to **`reduce`** in Python.

Both functions:
- Combine all elements of a collection into a **single value**
- Apply an operation cumulatively from left to right


### `fold` vs. `reduce`
- `fold` always requires an initial value

- `reduce` may or may not require one (depending on usage)

Because `fold` starts with an explicit initial value, it is:

- safer

- more predictable

- less error-prone for empty collections


```
from functools import reduce

result = reduce(lambda x, item: x + item, [1, 2, 3], 0)
print(result)  # 6
```

````kotlin
// The initial value is zero.
// The operation sums the initial value with every item in the list cumulatively.
println(listOf(1, 2, 3).fold(0, { x, item -> x + item })) // 6
````

If a lambda expression is passed as the **last parameter** of a function, the lambda can be written **outside** the parentheses.

````kotlin
// Alternatively, in the form of a trailing lambda
println(listOf(1, 2, 3).fold(0) { x, item -> x + item })  // 6
````
