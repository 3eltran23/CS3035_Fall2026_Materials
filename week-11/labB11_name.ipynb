### Provided list utilities
These utilities are supplied so you can focus on the stack. The `else` branches support notebook kernels that do not recognize exhaustiveness for these sealed-class declarations. The recursive factory and `foldRight` are intended for classroom-sized lists.


```kotlin
sealed class List<out A> {
    companion object {
        fun <A> of(vararg aa: A): List<A> {
            if (aa.isEmpty()) return Nil
            val tail = aa.sliceArray(1 until aa.size)
            return Cons(aa[0], of(*tail))
        }
    }
}

object Nil : List<Nothing>()

data class Cons<out A>(val head: A, val tail: List<A>) : List<A>()

fun <A> List<A>.tail(): List<A> =
    when (this) {
        is Cons -> this.tail
        is Nil -> throw IllegalStateException("Nil cannot have a tail")
        else -> error("Unexpected List subtype")
    }

fun <A, B> List<A>.foldRight(z: B, f: (A, B) -> B): B =
    when (this) {
        is Cons -> f(head, tail.foldRight(z, f))
        is Nil -> z
        else -> error("Unexpected List subtype")
    }

fun <A> List<A>.length(): Int =
    foldRight(0) { _, acc -> 1 + acc }
```
