````kotlin

fun <A, B> map(list: List<A>, f: (A) -> B): List<B> =
    when (list) {
        is Nil -> Nil
        is Cons-> Cons(f(list.head), map(list.tail, f))
        else -> error("Unexpected")
    }

````

````kotlin
fun <A> setHead(xs: List<A>, x: A): List<A> =
    when (xs) {
        is Nil ->
            throw IllegalStateException(
                "Cannot replace `head` of a Nil list"
            )
        is Cons -> Cons(x, xs.tail)
        else -> error("Unexpected List subtype: $xs")
    }
````

````kotlin
sealed class Tree<out A>

data class Leaf<A>(val value: A) : Tree<A>()

data class Branch<A>(
    val left: Tree<A>,
    val right: Tree<A>
) : Tree<A>()

fun <A> size(t: Tree<A>): Int =
    when (t) {
        is Leaf -> 1
        is Branch -> size(t.left) + size(t.right)
        else -> error("Unexpected Tree subtype: $t")
    }


fun <A> insert(t: Tree<A>, x: A): Tree<A> =
    when (t) {
        is Leaf -> Branch(t, Leaf(x))
        is Branch ->
            if (size(t.left) <= size(t.right))
                Branch(insert(t.left, x), t.right)
            else
                Branch(t.left, insert(t.right, x))
        else -> error("Unexpected Tree subtype: $t")
    }
````

````kotlin
val t0: Tree<Int> = Leaf(4)
val t1 = insert(t0, 3)
val t2 = insert(t1, 1)
````
